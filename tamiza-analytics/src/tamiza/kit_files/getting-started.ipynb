This notebook opens the sample you downloaded from Tamiza ({{submissions}}, made {{created}}). It runs here, with the files in this folder, and never contacts the server.

Run the cells from the top. Unlike the cells in Tamiza's editor, variables stay in memory from one cell to the next.

{{script1}}

```python
import tamiza_sdk as tz

GROUP = "_version"  # change to a question of the clean "main" table, such as "municipality"

ctx = tz.load_context()  # the ctx your script receives
raw = tz.load_tables("raw")  # the tables script 1 receives

try:
    cleaned = tz.load_tables("clean")  # the tables script 2 and the indicators receive
except tz.ScriptError as error:
    print(error)
    cleaned = raw  # carry on with the flattened tables

for name, table in cleaned.items():
    print(f"{name}: {len(table):,} rows, {len(table.columns)} columns")
```

## Look at the data

`ctx.show(value, name)` displays a value under a name, as it does in a test on a sample.

```python
main = cleaned["main"]
ctx.show(main.head(), "main, first rows")
ctx.show(main.dtypes.astype(str), "column types")
```

```python
# Try the calculation here, where you can look at each step.
counts = main.groupby(GROUP, dropna=False).size().rename("value").reset_index()
counts
```

## Write the script

A script is a Python file with one function. This cell writes an indicator to `households_by_group.py`; edit the file in Jupyter, or change the text below, and run the cell again.

```python
from pathlib import Path

NAME = "households_by_group"  # the indicator's name in Tamiza: lowercase letters, digits and underscores

Path(f"{NAME}.py").write_text(
    """
GROUP = "_version"  # change to a question of cleaned["main"]


def compute(cleaned, ctx):
    main = cleaned["main"]
    ctx.show(main, "main")
    return main.groupby(GROUP, dropna=False).size().rename("value").reset_index()
""".lstrip(),
    encoding="utf-8",
)
```

## Check it

`check` runs the file the way the server does, in a fresh namespace, with the same contract and validation, and returns the table it would publish. Names you defined only in this notebook are not visible to the file.

```python
# tables=cleaned gives the script the tables loaded above; leave it out to use the downloaded clean tables.
result = tz.check(f"{NAME}.py", "indicator", name=NAME, tables=cleaned)
result
```
